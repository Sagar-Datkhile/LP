# Practical 2: Email Spam Classification

**Aim:** Classify emails as spam (1) or not spam (0) using SVM and K-Nearest Neighbors (KNN).

**Dataset required:** `emails.csv` (place it in the same folder as this notebook).

```python
# Step 1: Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### Step 2: Load and view the dataset

```python
df = pd.read_csv("emails.csv")
print(df.head())
print("Missing values:\n", df.isnull().sum().sum())
```

### Step 3: Select input and output
This follows the notebook's column selection: columns 1 through 3000 are word-frequency features; the last column is the target label.

```python
X = df.iloc[:, 1:-1]
y = df.iloc[:, -1]
print('Input shape:', X.shape)
print('Output shape:', y.shape)
```

### Step 4: Split the data
75% training data and 25% testing data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

### Step 5: Support Vector Machine (SVM)

```python
svm = SVC(C=1.0, kernel='rbf', gamma='auto')
svm.fit(X_train, y_train)
svm_pred = svm.predict(X_test)

print('SVM Accuracy:', accuracy_score(y_test, svm_pred))
print('SVM Confusion Matrix:\n', confusion_matrix(y_test, svm_pred))
print('SVM Classification Report:\n', classification_report(y_test, svm_pred, zero_division=0))
```

### Step 6: K-Nearest Neighbors (KNN)

```python
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)

print('KNN Accuracy:', accuracy_score(y_test, knn_pred))
print('KNN Confusion Matrix:\n', confusion_matrix(y_test, knn_pred))
print('KNN Classification Report:\n', classification_report(y_test, knn_pred, zero_division=0))
```

### Step 7: Standardize data and compare K values
Scaling is especially useful for distance-based KNN. The scaler is fitted only on training data.

```python
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

for k in [1, 3, 5]:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_s, y_train)
    pred = model.predict(X_test_s)
    print(f'K = {k}, Accuracy = {accuracy_score(y_test, pred):.4f}')
```

### Result / Output explanation
Accuracy is the fraction of correct predictions. The confusion matrix compares actual and predicted classes. The classification report shows precision, recall and F1-score. Exact values depend on `emails.csv`.
